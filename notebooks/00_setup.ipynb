# 00 — Setup & Authentication

Run this notebook first to verify your environment is configured correctly.

**What this does:**
1. Checks that `.env` exists with your Brain credentials
2. Authenticates with WorldQuant Brain
3. Confirms the API is reachable and your account is active

**Before running:** copy `.env.example` to `.env` and fill in your credentials.

```python
import sys
from pathlib import Path

# Add repo root to path so we can import wq/
repo_root = Path.cwd().parent
if str(repo_root) not in sys.path:
    sys.path.insert(0, str(repo_root))
```

```python
# Check .env file exists
env_path = repo_root / '.env'
if not env_path.exists():
    raise FileNotFoundError(
        f'.env not found at {env_path}\n'
        'Copy .env.example to .env and fill in your Brain credentials.'
    )
print(f'.env found: {env_path}')
```

```python
# Check credentials are set (without printing them)
from dotenv import load_dotenv
import os

load_dotenv(env_path)
username = os.environ.get('WQ_USERNAME', '')
password = os.environ.get('WQ_PASSWORD', '')

assert username, 'WQ_USERNAME is not set in .env'
assert password, 'WQ_PASSWORD is not set in .env'
print(f'Credentials loaded for: {username}')
```

```python
# Authenticate with WorldQuant Brain
from wq.client import get_client

client = get_client()
print('Authentication successful!')
```

```python
# Confirm API is working by checking account details
resp = client.get('/users/self')
if resp.status_code == 200:
    user_info = resp.json()
    print('Account info:')
    for k, v in user_info.items():
        if k.lower() not in ('password', 'token', 'secret'):
            print(f'  {k}: {v}')
else:
    print(f'Could not fetch account info (status {resp.status_code}) — but auth worked fine.')
```

```python
# Check data directory is writable
data_dir = repo_root / 'data'
data_dir.mkdir(exist_ok=True)
test_file = data_dir / '.write_test'
test_file.write_text('ok')
test_file.unlink()
print(f'data/ directory is writable: {data_dir}')
```
